       Mathematical formulation of one-way ANOVA assumptions applied to binary dependent for deriving t-stat.
      </td>
      <td style="padding: 10px;">
        In a one-way ANOVA framework with a binary dependent variable, the data are divided into several groups, each with a certain probability of success. Each observation in a given group follows a Bernoulli distribution, meaning it can take only two possible values, typically coded as zero or one. The model assumes that the expected value of the dependent variable within each group corresponds to the group’s probability of success, while the variance depends on this same probability. Although the assumption of normal and homoskedastic errors is not exactly true for binary data, the central limit theorem ensures that the sample proportions are approximately normally distributed when sample sizes are large. Under the null hypothesis that all group proportions are equal, the variance is the same across groups, which allows the standard ANOVA or t-test framework to be applied. For two groups, the one-way ANOVA model reduces to the familiar two-sample t-test for comparing proportions. The test statistic measures the standardized difference between the two sample proportions, using a pooled estimate of the common variance under the null hypothesis. When the assumption of equal variances is relaxed, Welch’s version of the t-test can be used, which adjusts the standard error and degrees of freedom accordingly. For more than two groups, the ANOVA F-statistic compares the variability of group means (or proportions) around the overall mean to the variability within groups. When the dependent variable is binary, this F-test is asymptotically equivalent to the Pearson chi-square test for equality of proportions across multiple groups. Thus, applying one-way ANOVA to binary data conceptually yields the same inference as tests on proportions, provided sample sizes are sufficiently large and the variances across groups do not differ too much. In small samples or when group probabilities differ substantially, it is more appropriate to use heteroskedastic-robust methods or a binomial generalized linear model, such as a logistic regression, to formally test for equality across groups.

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
y_true = y_train  # Actual binary outcomes (0 or 1)
y_pred_prob = lpm_model2.predict(X_train)  # Predicted probabilities
cutoffs = np.linspace(0, 1, 101)

sensitivity = []
specificity = []

for c in cutoffs:
    y_pred = (y_pred_prob >= c).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    sens = tp / (tp + fn) if (tp + fn) > 0 else 0  # True Positive Rate
    spec = tn / (tn + fp) if (tn + fp) > 0 else 0  # True Negative Rate
    sensitivity.append(sens)
    specificity.append(spec)

plt.figure(figsize=(10, 6))
plt.plot(cutoffs, sensitivity, label="Sensitivity (True Positive Rate)", color="red", linewidth=2)
plt.plot(cutoffs, specificity, label="Specificity (True Negative Rate)", color="blue", linewidth=2)
plt.axvline(0.5, color="gray", linestyle="--", label="Cutoff = 0.5")

plt.title("Sensitivity and Specificity with Respect to Probability Cutoff (LPM)", fontsize=14)
plt.xlabel("Probability Cutoff", fontsize=12)
plt.ylabel("Rate", fontsize=12)
plt.legend(loc="center right", fontsize=10)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

neuron_y_gap = 50
layer_positions = {}
node_id = 0

def add_node(label, x, y, color="#DDEEFF", size=18):
    global node_id
    nid = f"n{node_id}"
    node_id += 1
    net.add_node(nid, label=label, x=x, y=y, physics=False, color=color, shape="dot", size=size)
    return nid

prev_neurons = []
for li, layer in enumerate(model.layers):
    x = li * layer_x_gap

    if isinstance(layer, layers.InputLayer):
        n_inputs = layer.input_shape[-1]
        neurons = []
        for ni in range(n_inputs):
            y = (ni - n_inputs/2) * neuron_y_gap
            neurons.append(add_node(f"in{ni}", x, y, color="#A7F3D0", size=14))
        layer_positions[layer.name] = (x, 0)
        # No incoming edges to inputs
    elif isinstance(layer, layers.Dense):
        n_units = layer.units
        neurons = []
        for ni in range(n_units):
            y = (ni - n_units/2) * neuron_y_gap
            color = "#BFDBFE" if "Output" not in layer.name else "#FDE68A"
            neurons.append(add_node(f"{layer.name}:{ni}", x, y, color=color, size=14))
        layer_positions[layer.name] = (x, 0)
        # fully connect previous layer to this one
        for p in prev_neurons:
            for q in neurons:
                net.add_edge(p, q, arrows="to")
    else:
        # For unsupported layer types, draw a single box node
        single = add_node(layer.name, x, 0, color="#E5E7EB", size=20)
        neurons = [single]
        for p in prev_neurons:
            net.add_edge(p, single, arrows="to")

    prev_neurons = neurons

for name, (x, _) in layer_positions.items():
